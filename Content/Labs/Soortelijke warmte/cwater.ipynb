````python
# Hier de data en de analyse
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
data_T = np.array([20.5, 20.9, 21.2, 21.6, 21.9, 22.3, 22.6, 22.9, 23.3, 23.6, 23.9, 24.3, 24.6, 24.9, 25.2, 25.6, 25.9, 26.2, 26.5, 26.8, 27.2, 
                  27.5, 27.7, 28.0, 28.3, 28.7, 28.9, 29.2, 29.6, 29.8, 30.2, 30.4, 30.8, 31.0, 31.3, 31.7, 32.0, 32.3, 32.6, 32.8, 33.2])
data_T2 = np.array([24.7, 25.0, 25.2, 25.3, 25.5, 25.7, 25.9, 26.0, 26.1, 26.2, 26.5, 26.7, 26.9, 27.0, 27.1, 27.3, 27.4, 27.6, 27.7, 27.8, 28.0, 
                  28.1, 28.2, 28.4, 28.6, 28.9, 29.0, 29.2, 29.3, 29.5, 29.5, 29.7, 29.8, 30.0, 30.1, 30.3, 30.4, 30.5, 30.7,30.8,31.0])
t = np.arange(0,601,15)
dt = 15
m = (641.6 - 267.2)/1000
m2 = (1043.7 - 267.2)/1000
P1 = 19.6* 2.02
P2 = 19.6 * 2.03
def soortelijk_w(t, T0, c, P, m):
    return T0 + (P / (m * c)) * t

popt1, _ = curve_fit(lambda t, T0, c: soortelijk_w(t, T0, c, P1, m1),t, data_T)

T0_1, c1 = popt1
print(popt1)

popt2, _ = curve_fit(lambda t, T0, c: soortelijk_w(t, T0, c, P2, m2),t, data_T2)

T0_2, c2 = popt2
print(popt2)
plt.plot(t, data_T, 'o', label='T1 data')
plt.plot(t, soortelijk_w(t, T0_1, c1, P1, m1), label='T1 fit')

plt.plot(t, data_T2, 'o', label='T2 data')
plt.plot(t, soortelijk_w(t, T0_2, c2, P2, m2), label='T2 fit')

plt.plot(t,data_T)
plt.plot(t,data_T2)
plt.xlabel("Tijd (s)")
plt.ylabel("Temperatuur (°C)")
plt.legend()
plt.show()

print("c uit meting 1:", c1, "J/(kg·K)")
print("c uit meting 2:", c2, "J/(kg·K)")
print("gemiddelde c:", (c1 + c2)/2)
````

# Discussie en conclusie

De gemeten waarde van c van water is 5029 J/(kg·K). Deze waarde is aanzienlijk hoger dan de reele waarde van c van water, 4186 J/(kg·K), wij denken dat dit verschil kan worden verklaard door een energieverlies naar de buitenlucht. Door deze uitwisseling lijkt het alsof er meer energie nodig is om het water een kelvin op te warmen. De proef zou dus beter werken als de omgevingstemperatuur mee stijgt met de watertemperatuur.
